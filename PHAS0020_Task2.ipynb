### Creating a function to map a 2D surface

Having sucessfully imported our libraries, we can now proceed to the first part of the code. Before we start our work with the solar images, we will need to define a function that will plot a 2D surface of a sphere onto our images of the Sun. We define this function below as "mapping", and it coverts cylindrical polar coordinates into x and z-coords that will be used to construct a 2D model plotted on our solar images. Our function will use the following conversion from cylindrical polar coordinates to cartesian:
$$ x = r \cos(\theta)\cos(\phi) \\$$
$$y = r \cos(\theta)\sin(\phi) \\ $$
$$z = r \sin(\theta)$$
Please note that we will only be returning x and z-coords in the function below since we are creating a 2D model.

```python
### STUDENT CODE CELL - COMPLETE for Task 1.2 ###

# Define a function that takes in 3 inputs (i.e. r,phi,theta) and returns the two desired cartesian coordinates x and z
def mapping (r, phi, theta):
    """This function will convert cylindrical polar into cartesian coords and will be used for mapping later in the task
    Inputs:
    r = radius 
    theta= polar angle 
    phi = azimuthal angle
    Returns:
    x = x-coord
    z = z-coord
    """

    #compute the cartesian coords
    x = r*np.cos(theta)*np.cos(phi)
    y = r*np.cos(theta)*np.sin(phi)
    z = r*np.sin(theta)

    return x,z
```

### Locating sunspots 

We adapt the same procedure when it comes to mapping our sunspots as before (markers, meridians, latitudes). We find the angles of $\phi$ and $\theta$ for all the sunspots. 

**A note on the errors of these measuremenets** : We can make a note on the uncertainities here. As the sunspots were plotted manually, we estimate the uncertainity by changing our pixel values and see that our uncertainty is 2 pixels for both $\Delta x$ and $\Delta z$ and  as below that value we cannot identify that there is a change in the marker we assign to our sunspot. We will need to propagate this error by the standard procedure of finding errors of one variable. The error on $\phi$ is hence found by:
$$
\Delta\phi = \frac{\partial}{\partial x} * arccos (\frac{xx - xc}{rr}) * \Delta x
$$
$$
\Delta\phi = \frac{2}{rr * \sqrt{1 - \frac{(xx-xc)^2}{rr^2}}}
$$
And for the error on $\theta$:
$$
\Delta\theta = \frac{\partial}{\partial x} * arcsin (\frac{zz - zc}{rr}) * \Delta z
$$
$$
\Delta\theta = \frac{2}{rr * \sqrt{1 - \frac{(zz-zc)^2}{rr^2}}}
$$

We print the angles and their errors above our graphs with the meridians and latitude lines. This is done for clarity in order to understand which measurement corresponds to which graph instead of denoting them in a spearate markdown cell to avoid being repetition.

### Calculating angular rotations

We will now find the angular rotation by plotting the values of $\phi$ against time for each sunspot.

```python
### STUDENT CODE CELL - COMPLETE for Task 2.5 & 2.6 ###

# In this cell calculate the Angular Rotation of each sunspot by perfoming a linear fit of the values of phi vs time.

#create arrays of phi for each sunspot 
phis1 = np.array ([phi_sunspot11, phi_sunspot12, phi_sunspot13])
phis2 = np.array ([phi_sunspot21, phi_sunspot22, phi_sunspot23])
phis3 = np.array ([phi_sunspot31, phi_sunspot32, phi_sunspot33])
```

By fitting a linear fit, we will be able to obtain the gradient and the uncertainty of the fit. This represents the angular rotation.

```python
#finding the fit using np.polyfit()
degree = 1
xline = np.array ([0,50])

#sunspot1
coeffs_fit1, errors_fit1 = np.polyfit(times, phis1, degree, cov="unscaled")
m1, c1 = coeffs_fit1
dm1, dc1 = np.sqrt(np.diag(errors_fit1)) 
yline1 = m1 *xline +c1

#sunspot 2
coeffs_fit2, errors_fit2 = np.polyfit(times, phis2, degree, cov="unscaled")
m2, c2 = coeffs_fit2
dm2, dc2 = np.sqrt(np.diag(errors_fit2)) 
yline2 = m2 *xline +c2

#sunspot 3
coeffs_fit3, errors_fit3 = np.polyfit(times, phis3, degree, cov="unscaled")
m3, c3 = coeffs_fit3
dm3, dc3 = np.sqrt(np.diag(errors_fit3)) 
yline3 = m3 *xline +c3

#loading data for error bars
err_1 = np.array ([err_phi_11, err_phi_12, err_phi_13])
err_2 = np.array ([err_phi_21, err_phi_22, err_phi_23])
err_3 = np.array ([err_phi_31, err_phi_32, err_phi_33])


#plotting
plt.errorbar(times, phis1, yerr = err_1, capsize = 5, fmt='b.')
plt.plot (xline, yline1, color = "blue", label = "sunspot 1")
plt.errorbar (times, phis2, yerr = err_2, capsize = 5, fmt='g.')
plt.plot (xline, yline2, color = "green", label = "sunspot 2")
plt.errorbar (times, phis3, yerr = err_3, capsize = 5, fmt='r.')
plt.plot (xline, yline3, color = "red", label = "sunspot 3")

plt.title("Phi VS Time for each sunspot")
plt.xlabel("Time [h]")
plt.ylabel("Phi [radians]")
plt.legend ()

#Adding text to the plot
plt.text(0, 0.8,f"sunspot1: m = {m1:.5f} +- {dm1:.5f}")
plt.text(0, 0.7,f"sunspot2: m = {m2:.5f} +- {dm2:.5f}")
plt.text(0, 0.6,f"sunspot3: m = {m3:.5f} +- {dm3:.5f}")
```

By plotting our graph, we see that there is a linear dependence between the changing angle of $\phi$ over time. We see that our data points (sunspots on each image) do not fit precisely into a linear model, and do not pass through the error bars of our data points. This implies that we underestimated our errors when placing merridians, suggesting the error in measurement was actually larger than $\Delta x$ = 2. This is also implied by the error in our comupted values of the gradient, since they are greater than the values of the gradient itself.

### Comparison to published models

<div class="alert alert-success">
If you google "Solar Rotation" you will find on wikipedia that a simple formula can be used (from Snodgrass, H.; Ulrich, R. (1990). "Rotation of Doppler features in the solar photosphere". Astrophysical Journal. 351: 309–316.)
for the rotation of the Sun.
This is copied here for your convenience.

$$ \omega = A + B \sin^2(\varphi) + C \sin^4(\varphi) $$

where 

$ \omega $ is the angular velocity in degrees per day, 
$ \varphi $ is the solar latitude, A is angular velocity at the equator, and B, C are constants controlling the decrease in velocity with increasing latitude. 

The values of A, B, and C differ depending on the techniques used to make the measurement, as well as the time period studied. A current set of accepted average values[R] is:

$ A= 14.713 \pm 0.0491 \mbox{   } ^{\circ} /day $<br>
$ B= −2.396 \pm 0.188  \mbox{   } ^{\circ}/day  $<br>
$ C= −1.787 \pm 0.253  \mbox{   } ^{\circ}/day  $<br>

Note that there are alternative formulations (one such that you can check is the Ulrich et al. 1988 which has different units - if you have time consider looking it up).
</div>
